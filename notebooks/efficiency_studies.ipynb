# Efficiency Studies

Use RapidSim-generated samples of $D^0\to K^0_S\pi^+\pi^-$ and $D^0\to\pi^+\pi^-$ to determine the possibility of correcting the trigger efficiency of the former with the latter:

$$
\begin{align}
A_{\text{raw}}(D^{*+}\to D^0(\to K^0_S\pi^+\pi^-)\pi_s^+) &= A_{\text{HLT1}}(D^0\to K^0_S\pi^+\pi^-) + A_{\text{det}}(K^0_S) + A_{\text{det}}(\pi^+\pi^-) \nonumber \\
&+ A_{\text{prod}}(D^{*+}) + A_{\text{det}}(\pi_s^+) \nonumber\\
&+ A_{\text{CP}}^{\text{dir}}(D^0\to K^0_S\pi^+\pi^-) \\
A_{\text{raw}}(D^{*+}\to D^0(\to\pi^+\pi^-)\pi_s^+) &= A_{\text{HLT1}}(D^0\to\pi^+\pi^-) + A_{\text{det}}(\pi^+\pi^-) \nonumber\\
&+ A_{\text{prod}}(D^{*+}) + A_{\text{det}}(\pi_s^+) \nonumber\\
&+ A_{\text{CP}}^{\text{dir}}(D^0\to\pi^+\pi^-)
\end{align}
$$

We can assume $A_{\text{CP}}^{\text{dir}}(D^0\to K^0_S\pi^+\pi^-)\approx 0$, while $A_{\text{CP}}^{\text{dir}}(D^0\to\pi^+\pi^-) = (2.30 \pm 0.56) \times 10^{-3}$.

The HLT1 asymmetry, aside the $D^0$ mass window, can be factorised as 
$$
\begin{align}
A_{\text{HLT1}}(D^0\to K^0_S\pi^+\pi^-) &\approx A_{\text{HLT1}}(K^0_S) + A_{\text{HLT1}}(\pi^+\pi^-) \\
A_{\text{HLT1}}(D^0\to \pi^+\pi^-) &\approx A_{\text{HLT1}}(\pi^+\pi^-)
\end{align}
$$
in the region where the asymmetries are up to a few percents.

And the raw asymmetries become
$$
\begin{align}
A_{\text{raw}}(D^{*+}\to D^0(\to K^0_S\pi^+\pi^-)\pi_s^+) &= A_{\text{HLT1}}(K^0_S) + A_{\text{HLT1}}(\pi^+\pi^-) + A_{\text{det}}(K^0_S) + A_{\text{det}}(\pi^+\pi^-) \nonumber\\
&+ A_{\text{prod}}(D^{*+}) + A_{\text{det}}(\pi_s^+) \\
A_{\text{raw}}(D^{*+}\to D^0(\to\pi^+\pi^-)\pi_s^+) &= A_{\text{HLT1}}(\pi^+\pi^-) + A_{\text{det}}(\pi^+\pi^-) \nonumber\\
&+ A_{\text{prod}}(D^{*+}) + A_{\text{det}}(\pi_s^+)
\end{align}
$$

Implying that 
$$
\begin{align}
A_{\text{raw}}(D^{*+}\to D^0(\to K^0_S\pi^+\pi^-)\pi_s^+) &= A_{\text{raw}}(D^{*+}\to D^0(\to\pi^+\pi^-)\pi_s^+) + A_{\text{HLT1}}(K^0_S) + A_{\text{det}}(K^0_S)
\end{align}
$$
when the kinematic distributions of the $D^{*+}$, $\pi^+_s$, $\pi^+$, and $\pi^-$ are equalised.

```python
from analysis_helpers.plotting import plot_hists
import matplotlib.pyplot as plt
```

```python
vars_to_plot = {
    'Dst_PT': {'name': r'$p_T(D^*)$', 'unit': 'GeV/c', 'range': (0, 20)}, 
    'Dst_P': {'name': r'$p(D^*)$', 'unit': 'GeV/c', 'range': (0, 500)}, 
    'Dst_eta': {'name': r'$\eta(D^*)$', 'unit': '', 'log': True, 'range': (1.5, 6)}
    }
for var, props in vars_to_plot.items():
    d2pipi_values = d2pipi['DecayTree'][var].array(library="np")
    d2kspipi_values = d2kspipi['DecayTree'][var].array(library="np")
    fig, ax = plot_hists([d2pipi_values, d2kspipi_values], legend=[r'$D^*\to D^0(\to\pi\pi)\pi^+$', r'$D^*\to D^0(\to K_S^0\pi\pi)\pi^+$'], name=props['name'], unit=props.get('unit', ''), density=True, bins=50, range=props.get('range', None), log=props.get('log', False))
    ax.legend()
```
