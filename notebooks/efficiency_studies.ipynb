# Efficiency Studies

Use RapidSim-generated samples of $D^0\to K^0_S\pi^+\pi^-$ and $D^0\to\pi^+\pi^-$ to determine the possibility of correcting the trigger efficiency of the former with the latter:

$$
\begin{align}
A_{\text{raw}}(D^{*+}\to D^0(\to K^0_S\pi^+\pi^-)\pi_s^+) &= A_{\text{HLT1}}(D^0\to K^0_S\pi^+\pi^-) + A_{\text{det}}(K^0_S) + A_{\text{det}}(\pi^+\pi^-) \nonumber \\
&+ A_{\text{prod}}(D^{*+}) + A_{\text{det}}(\pi_s^+) \nonumber\\
&+ A_{\text{CP}}^{\text{dir}}(D^0\to K^0_S\pi^+\pi^-) \\
A_{\text{raw}}(D^{*+}\to D^0(\to\pi^+\pi^-)\pi_s^+) &= A_{\text{HLT1}}(D^0\to\pi^+\pi^-) + A_{\text{det}}(\pi^+\pi^-) \nonumber\\
&+ A_{\text{prod}}(D^{*+}) + A_{\text{det}}(\pi_s^+) \nonumber\\
&+ A_{\text{CP}}^{\text{dir}}(D^0\to\pi^+\pi^-)
\end{align}
$$

We can assume $A_{\text{CP}}^{\text{dir}}(D^0\to K^0_S\pi^+\pi^-)\approx 0$, while $A_{\text{CP}}^{\text{dir}}(D^0\to\pi^+\pi^-) = (2.30 \pm 0.56) \times 10^{-3}$.

The HLT1 asymmetry, aside the $D^0$ mass window, can be factorised as 
$$
\begin{align}
A_{\text{HLT1}}(D^0\to K^0_S\pi^+\pi^-) &\approx A_{\text{HLT1}}(K^0_S) + A_{\text{HLT1}}(\pi^+\pi^-) \\
A_{\text{HLT1}}(D^0\to \pi^+\pi^-) &\approx A_{\text{HLT1}}(\pi^+\pi^-)
\end{align}
$$
in the region where the asymmetries are up to a few percents.

And the raw asymmetries become
$$
\begin{align}
A_{\text{raw}}(D^{*+}\to D^0(\to K^0_S\pi^+\pi^-)\pi_s^+) &= A_{\text{HLT1}}(K^0_S) + A_{\text{HLT1}}(\pi^+\pi^-) + A_{\text{det}}(K^0_S) + A_{\text{det}}(\pi^+\pi^-) \nonumber\\
&+ A_{\text{prod}}(D^{*+}) + A_{\text{det}}(\pi_s^+) \\
A_{\text{raw}}(D^{*+}\to D^0(\to\pi^+\pi^-)\pi_s^+) &= A_{\text{HLT1}}(\pi^+\pi^-) + A_{\text{det}}(\pi^+\pi^-) \nonumber\\
&+ A_{\text{prod}}(D^{*+}) + A_{\text{det}}(\pi_s^+)
\end{align}
$$

Implying that 
$$
\begin{align}
A_{\text{raw}}(D^{*+}\to D^0(\to K^0_S\pi^+\pi^-)\pi_s^+) &= A_{\text{raw}}(D^{*+}\to D^0(\to\pi^+\pi^-)\pi_s^+) + A_{\text{HLT1}}(K^0_S) + A_{\text{det}}(K^0_S)
\end{align}
$$
when the kinematic distributions of the $D^{*+}$, $\pi^+_s$, $\pi^+$, and $\pi^-$ are equalised.

## Create a Dataframe to apply the selection

To determine whether a candidate is selected, we use dataframes with variables calculated out of the RapidSim output.

```python
# ['hh_PT', 'hh_VERTEX_CHI2', 'hh_ETA', 'hh_DOCA', 'h1_PT', 'h2_PT', 'h1_IP', 'h2_IP', 'hh_CTAU', 'hh_MASS', 'hh_Z', 'PV_Z']
vars_to_use  = [f'{x}_P{y}' for x in ['pip', 'pim'] for y in ['X', 'Y', 'Z']]
vars_to_use += [f'{x}_IP' for x in ['pip', 'pim']]
vars_to_use += ['D0_vtxX', 'D0_vtxY', 'D0_vtxZ', 'Dst_vtxX', 'Dst_vtxY', 'Dst_vtxZ', 'D0_eta', 'D0_M']
df = d2pipi['DecayTree'].arrays(vars_to_use, library="pd")
df.head()
```

```python
from analysis_helpers.kinematics import pseudorapidity, doca, slope, ctau, mass
import numpy as np
df['hh_PT'] = 1000.*( (df['pip_PX']+df['pim_PX'])**2 + (df['pip_PY']+df['pim_PY'])**2)**0.5
df['hh_VERTEX_CHI2'] = np.zeros(len(df))  # Placeholder, need to compute from vertex fit
df['hh_ETA'] = pseudorapidity( df['pip_PX']+df['pim_PX'], df['pip_PY']+df['pim_PY'], df['pip_PZ']+df['pim_PZ'] )
df['hh_DOCA'] = doca( df['D0_vtxX'], df['D0_vtxY'], df['D0_vtxZ'], slope(df['pip_PX'], df['pip_PZ']), slope(df['pip_PY'], df['pip_PZ']), 
                      df['D0_vtxX'], df['D0_vtxY'], df['D0_vtxZ'], slope(df['pim_PX'], df['pim_PZ']), slope(df['pim_PY'], df['pim_PZ']) )
df['h1_PT'] = 1000.*(df['pip_PX']**2 + df['pip_PY']**2)**0.5
df['h2_PT'] = 1000.*(df['pim_PX']**2 + df['pim_PY']**2)**0.5
df['h1_IP'] = df['pip_IP']
df['h2_IP'] = df['pim_IP']
df['D0_DL'] = np.sqrt( (df['D0_vtxX'] - df['Dst_vtxX'])**2 + (df['D0_vtxY'] - df['Dst_vtxY'])**2 + (df['D0_vtxZ'] - df['Dst_vtxZ'])**2 )
df['hh_CTAU'] = ctau( df['D0_DL'], 1864.83, df['hh_PT'] )
df['pip_E'] = np.sqrt(df['pip_PX']**2 + df['pip_PY']**2 + df['pip_PZ']**2 + 0.13957**2)
df['pim_E'] = np.sqrt(df['pim_PX']**2 + df['pim_PY']**2 + df['pim_PZ']**2 + 0.13957**2)
df['hh_MASS'] = 1000.*mass( df['pip_PX'] + df['pim_PX'], df['pip_PY'] + df['pim_PY'], df['pip_PZ'] + df['pim_PZ'], df['pip_E'] + df['pim_E'] )
df['hh_Z'] = df['D0_vtxZ']
df['PV_Z'] = df['Dst_vtxZ']  # Assuming PV is at
```
